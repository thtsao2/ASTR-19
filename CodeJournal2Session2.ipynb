```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2 * np.pi, 100)

def decay_func(x, a, b):
    return a * np.exp(-b * x)

a_val = 5
b_val = 0.5
```

```python
y = decay_func(x, a_val, b_val)

plt.figure(figsize=(8, 5))
plt.plot(x, y, linewidth = 2)

plt.xlabel('x (0 to pi)')
plt.ylabel('y')
plt.title('Exponential Decay Plot')
plt.grid(True, linestyle='--', alpha=0.7)

plt.savefig('decay_plot.pdf')

plt.show()
```

```python
sigma = np.linspace(2.0, 0.2, len(x))

noise = np.random.normal(0, sigma)
y_scattered = y + noise

plt.figure(figsize=(10,6))

plt.errorbar(x, y_scattered, yerr=sigma, fmt='o', color='red', elinewidth=1, capsize=3, label='Scattered Data')

plt.plot(x, y, color='black', linestyle='--', alpha=0.7, label='Original')

plt.xlabel('x (time / distance)')
plt.ylabel('y (amp)')
plt.title('Exp Decay with Linearly Decreasing Uncertainty')
plt.grid(True, alpha=0.3)

plt.savefig('scattered_decay_plot.pdf')

plt.show()
```
