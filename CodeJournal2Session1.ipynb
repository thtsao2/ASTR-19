```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2 * np.pi, 100)

def decay_func(x, a, b):
    return a * np.exp(-b * x)

a_val = 5
b_val = 0.5
```

```python
y = decay_func(x, a_val, b_val)

plt.figure(figsize=(8, 5))
plt.plot(x, y, linewidth = 2)

plt.xlabel('x (0 to pi)')
plt.ylabel('y')
plt.title('Exponential Decay Plot')
plt.grid(True, linestyle='--', alpha=0.7)

plt.savefig('decay_plot.pdf')

plt.show()
```
