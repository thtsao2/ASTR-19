```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()

data = rng.rayleigh(scale=1.0, size=1000)
```

```python
plt.figure(figsize=(10,6))

plt.hist(data, bins=100, color='blue', alpha=0.7)

plt.xlabel("value")
plt.ylabel("frequency (normalized)")
plt.title("Histogram of 1000 Samples from a Rayleigh Distribution")

plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.savefig('rayleigh_histogram.pdf')

plt.show()
```
